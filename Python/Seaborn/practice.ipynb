### Using sales, draw a lineplot of date vs sales.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
print("Importing completed!")
```

```python
sales_data = pd.read_csv('sales.csv')
sales_data.head()
```

```python
sns.lineplot(data=sales_data, x='date', y='sales')
plt.show()
```

### From students, plot hours_studied vs marks and color by gender.

```python
students = pd.read_csv('student_dataset_complete.csv')
students.head()
```

```python
sns.lineplot(data=students, x='study_hours', y='Marks', hue='gender')
plt.show()
```

### Create a grid of scatterplots (hours_studied vs marks) faceted by subject.

```python
sns.relplot(kind='scatter', data=students, x='study_hours', y='Marks', col='subject')
plt.show()
```

### Plot a histogram of marks with a KDE overlay.

```python
sns.histplot(kde=True, data=students, x='Marks')
plt.show()
```

### Show the KDE curve for hours_studied.


```python
sns.kdeplot(data=students, x='study_hours')
plt.show()
```

### Display the count of records per subject.


```python
sns.barplot(data=students, x='subject', y='Marks', errorbar=None)
```

### Draw a regression line for hours_studied and marks.


```python
sns.regplot(data=students, x='study_hours', y='Marks')
plt.show()
```

### Create a pairplot for students' dataset. 


```python
students_marks = students[['time_study', 'Marks', 'study_hours', 'test_score']]

sns.pairplot(data=students_marks)
plt.show()
```

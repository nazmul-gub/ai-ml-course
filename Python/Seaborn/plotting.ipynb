```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
print("Importing completed!")
```

```python
student = pd.read_csv("sns_data.csv")
student.head()
```

### Lineplot using seaborn

```python
# axis level

sns.lineplot(data=student, x='week', y='attendance_rate', errorbar=None)
plt.show()
```

```python
sns.lineplot(data=student, x='week', y='attendance_rate', errorbar=None, hue='gender')
plt.show()
```

```python
#figure level

sns.relplot(kind='line', data=student, x = 'week' , y = 'attendance_rate',errorbar=None,hue='gender')#
plt.show()
```

```python
sns.relplot(kind ='line', data= student , x = 'week' , y = 'test_score',errorbar=None,hue='gender')
```

### exploring boxplot

```python
sns.set_theme(style='whitegrid')

tips = sns.load_dataset('tips')
sns.boxplot(x=tips["total_bill"])
plt.show()
```

```python
sns.boxplot(data=tips, x='day', y='total_bill')
plt.show()
```

```python
tips.head()
```

```python
sns.catplot(kind='box', data=tips, x='day', y='total_bill', hue='smoker', palette="pastel")
plt.show()
```

### violinplot

```python
sns.violinplot(data=tips['total_bill'])
plt.show()
```

```python
sns.violinplot(data=tips, x='day', y='total_bill')
plt.show()
```

```python
sns.violinplot(data=tips, x='day', y='total_bill', hue='smoker')
plt.show()
```

```python
sns.violinplot(data=tips, x='day', y='total_bill', hue='smoker', split=True, inner='quart')
plt.show()
```

### 2D line plot from pandas data

```python
df = pd.read_csv('enrollment_data.csv') 
df.head()
```

```python
plt.xlabel('Years') 
plt.ylabel('Count of Enrollment')
plt.title('Student Enrollment over 8 years')

plt.plot(df["Year"], df["Programming"])
plt.plot(df['Year'] ,df['Digital Marketing'])
```

```python
# adding labels using plt.legend() 
plt.xlabel('Years') 
plt.ylabel('Count of Enrollment')
plt.title('Student Enrollment over 8 years')

plt.plot(df["Year"], df["Programming"], label="programming")
plt.plot(df['Year'] ,df['Digital Marketing'], label="digtal marketing")
plt.legend()
```

```python
# adding some visualization 

plt.xlabel('Years') 
plt.ylabel('Count of Enrollment')
plt.title('Student Enrollment over 8 years')

plt.plot(df["Year"], df["Programming"], label="programming", color="green", linewidth=2, marker='o', markersize=5)
plt.plot(df['Year'] ,df['Digital Marketing'], label="digtal marketing", color="purple", linewidth=2, marker='o', markersize=5)


plt.legend()
plt.grid()
plt.show()
```

### Scatter Plot

### Bar Chart

```python
data = {
    'Student': [ 'B', 'C', 'D', 'E', 'F', 'G', 'H'],
    'Gender': [ 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male'],
    'StudyHours': [ 2, 5, 3, 7, 4, 8, 1],
    'Attendance': [ 60, 88, 70, 95, 80, 98, 55],
    'Grade': [ 'Fail', 'Pass', 'Fail', 'Pass', 'Pass', 'Pass', 'Fail']
}

df = pd.DataFrame(data)
```

```python
df
```

```python
# gender wise

gender_group = df.groupby('Gender').size()
print(gender_group.index)
print(gender_group.values)
```

```python
plt.bar(gender_group.index, gender_group.values, color=["pink", "skyblue"], edgecolor= "black")

plt.xlabel('Gender') 
plt.ylabel('Count') 
plt.title('Bar Chart of Gender among students')
plt.show()
```

```python
#  Bar chart on real data 

df = pd.read_csv('student_data.csv') 

df
```

```python
status = df.groupby("CompletionStatus").size()
print(status.index)
print(status.values)
```

```python
plt.bar(status.index , status.values, color=['green','orange','red'],edgecolor='black')
plt.xlabel('Completion Status') 
plt.ylabel('Count') 
plt.title('Bar Chart of Course completion')
plt.show() 
```

```python
location = df.groupby('Location').size()

plt.bar(location.index, location.values)

plt.xlabel('Location') 
plt.ylabel('Count') 
plt.title('Bar Chart of Student Location')
plt.show() 
```

### pie chart

```python
status = df.groupby('CompletionStatus').size() 

plt.pie(status, labels=status.index, autopct='%1.1f%%', shadow=True, explode=(0.05, 0, 0))
plt.show()
```

```python
location = df.groupby('Location').size()

plt.pie(location, shadow=True, labels=location.index)
plt.show()
```

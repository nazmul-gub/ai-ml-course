```python
import pandas as pd
import numpy as np

raw_df = pd.read_csv('employee_records.csv')
df = raw_df

df
```

```python
df.columns
```

```python
df.columns = df.columns.str.strip().str.lower()
df.columns
```

```python
df.info()
```

```python
# clean whitespace in columns

for i in df.select_dtypes(include=['object']).columns:
    df[i] = df[i].astype(str).str.strip()
```

```python
# normalize null values

null_values = null_values_list = [ None,np.nan,pd.NA,pd.NaT,np.inf,np.inf,"","NaN","null","N/A"]

for i in df.columns:
    if df[i].dtype == "object":
        df[i] = df[i].replace(to_replace= null_values, value=np.nan)
```

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eve", "Frank", "Grace", "Hannah","Sakib", "Nazmul"],
    "City": ["New York", "Los Angeles", "Newark", "Boston", "New Delhi", "Chicago", "New Orleans", "Houston","H Los Ang", "new Market"],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "Marketing", "Finance", "HR", "HR", "AI"],
    "Salary": [50000, 60000, 55000, 70000, 52000, 58000, 62000, 51000,70000, 100000]
}

df = pd.DataFrame(data)
df.head()
```

```python
df.loc[df['City'].str.contains("New")]
```

```python
# case insensivity

df.loc[df['City'].str.contains("New", case=False)]
```

```python
# starts with Los

df.loc[df['City'].str.contains(r"^Los")]
```

```python
#end with rk

df.loc[df['City'].str.contains(r"rk$")]
```

```python
# name starts with a vowel 

df.loc[df["Name"].str.contains(r"[EIOUN]")]
```

```python
# string contains New or Los

df.loc[df["City"].str.contains(r"New|Bos", case=False)]
```

### Adding New Colums

```python
df = pd.read_csv("student_data.csv")
df.head()
```

```python
# adding a constant column

df["Country"] = "Bangladesh"
df.head()
```

```python
df["Total Marks"] = df["Data Structure Marks"] + df["Algorithm Marks"] + df["Python Marks"]
df.head()
```

```python
print(df["Data Structure Marks"].dtype)
print(df["Data Structure Marks"].unique())
```

```python
df["Data Structure Marks"] = pd.to_numeric(df["Data Structure Marks"], errors="coerce")
```

```python
import numpy as np

df["A+ in DS"] = np.where(df["Data Structure Marks"] >= 80, "A+", "F")
df.head()
```

```python
df["Passed in DS"] = df["Data Structure Marks"] > 40
df.head()
```

```python
df["First Name"] = df["FullName"].str.split(" ").str[0]

df.head()
```

```python
df.to_csv("new_data.csv")
```

### Handling Duplicates Values

```python
import pandas as pd
data = {
    "Name": ["Alice", "Bob", "Charlie", "Alice", "David", "Bob"],
    "City": ["New York", "London", "Paris", "New York", "Tokyo", "London"],
    "Score": [85, 90, 78, 70, 95, 90]
}

df = pd.DataFrame(data)
df
```

```python
df.duplicated().sum()
```

```python
# delelting duplicates values 

df.drop_duplicates() 
```

```python
# permanent 

df.drop_duplicates(inplace= True) 
```

```python
df
```

```python
# name er basis e duplicate deletion 

df.drop_duplicates(subset="Name")
```

```python
# city
df.drop_duplicates(subset=['City']) 
```

```python
df.drop_duplicates(subset=['Name'],keep='last') 
```

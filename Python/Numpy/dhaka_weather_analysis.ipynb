## 30 days of weather data containing temperature, humidity, and rainfall.

```python
import numpy as np

# each row represents one day, in chronological order
# columns -> temperature (c), humidity(%), rainfall (mm)

weather = np.array([
    [31.2, 72,  0.0],
    [32.5, 68,  0.0],
    [33.1, 74,  0.0],
    [30.8, 81,  4.5],
    [29.7, 86, 12.0],
    [31.4, 79,  0.0],
    [34.0, 70,  0.0],
    [35.2, 66,  0.0],
    [34.6, 73,  0.0],
    [32.8, 78,  0.0],
    [31.9, 84,  2.0],
    [30.5, 88,  8.5],
    [29.8, 91, 15.0],
    [30.2, 87,  6.0],
    [32.1, 80,  0.0],
    [33.7, 75,  0.0],
    [34.8, 69,  0.0],
    [35.5, 64,  0.0],
    [36.1, 61,  0.0],
    [34.2, 72,  0.0],
    [33.0, 77,  0.0],
    [31.6, 83,  1.5],
    [30.9, 89,  7.0],
    [29.5, 92, 18.0],
    [30.7, 85,  5.0],
    [32.4, 79,  0.0],
    [33.8, 74,  0.0],
    [34.9, 67,  0.0],
    [35.7, 63,  0.0],
    [32.2, 81,  0.5]
]) 
```

#### exploring the dataset

```python
print(weather.dtype)
print(weather.size)
print(weather.shape)
print(weather.ndim)
```

```python
temperature = weather[:,1]

print(f"All Temperature Records: {temperature.ravel()}")

humidity = weather[:, 1]

print(f"All Humidity Records: {humidity.ravel()}")

rainfall = weather[:, 2]

print(f"All Rainfall Records: {rainfall.ravel()}")
```

```python
temperature_fahrenheit = (temperature*1.8) + 32

print(f"Temperature in Fahrenheit {temperature_fahrenheit.ravel()}")
```

#### weather statistics

```python
average_temp = np.mean(temperature)

print(f"Average temperature is: {average_temp}")

max_temp = np.max(temperature)

print(f"Maximum temperature is: {max_temp}")

min_temp = np.min(temperature)

print(f"Minimum temperature is: {min_temp}")
```

Investgating hot days

```python
hot_days = temperature[temperature > 34]

print(f"Hot days temperatures -> {hot_days}")

count_of_hot_days = len(hot_days)
print(f"There are {count_of_hot_days} hot days in this month.")
```

analyzing rainfalls

```python
rainy_days = rainfall[rainfall > 0]

total_rainy_days = len(rainy_days)
print(f"Total rainy days in this month : {total_rainy_days}")

rainy_days_greater_5 = len(rainfall[rainfall > 5]) 
print(f"Total rainy days greater than 5mm in this month : {rainy_days_greater_5}")

total_count_rainfall = np.sum(rainfall)
print(f"Total rainfall count : {total_count_rainfall} mm")

highest_rainfall = np.max(rainfall)
print(f"Highest rainfall of this month : {highest_rainfall} mm")
```

finding the days when temperature was above the overall average and rainfall was greater than zero...

```python
days = temperature[(temperature > average_temp) & (rainfall > 0)]

print(days)

# there are no days like that
```

#### classifying the weather

is temp > 34 then count that a hot day otherwise it is not a hot day

```python
classifying_days = np.where(temperature > 34, "Hot", "Not Hot")

print(classifying_days.ravel())
```

## Summary of Findings

The analysis of 30 weather observations showed an average temperature of **32.63°C**, with temperatures ranging from **29.5°C to 36.1°C**. Nine observations recorded temperatures of at least 34°C, representing 30% of the sample.

Rainfall occurred in 11 out of 30 observations, while the remaining 19 had no recorded rainfall. The total recorded rainfall was **80 mm**, and six observations received more than 5 mm. Observation 24 recorded both the lowest temperature and the highest rainfall, at 29.5°C and 18 mm, respectively.

One notable finding was that none of the rainy observations had temperatures above the overall mean of 32.63°C. This describes a pattern within the supplied dataset, but it does not establish a general relationship between rainfall and temperature.

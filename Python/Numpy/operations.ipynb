### slicing

```python
import numpy as np
```

```python
arr = np.array([[1,2,3],
                [4,5,6]])

print(f"Whole row: {arr[0]}")
print(f"Whole column: {arr[:,0]}")
print(f"Specific element: {arr[1,1]}")
```

### sorting

```python
arr_2d = np.array([[23, 24], [67,43], [1, -9]])

print(f"sorted by column: {np.sort(arr_2d, axis=0)}")
print(f"sorted by row: {np.sort(arr_2d, axis=1)}")
```

## filter

```python
nums = np.arange(11)
evens = nums[nums % 2 == 0]
print(evens)
```

filter with mask

```python
mask = nums > 5
print(nums[mask])
```

```python
idx = [0,2,3]
n = np.array([24,345,77,8,32, 652, -65, 54, 0])
print(n[idx])
```

```python
where_result = np.where(n > 50)
print(where_result)
print(n[where_result])
```

```python
conditional_array = np.where(n <= 0, n, "False")
print(conditional_array)
```

```python
a1 = np.arange(5)
a2 = np.arange(6, 11)

a1_a2_sum = a1+a2
print(a1_a2_sum)

add = np.concatenate((a1, a2))
print(add)
```

array compatibility

```python
nums1 = np.array([12, 45, 7, 23, 89, 54, 3, 18, 90, 31])
nums2 = np.array([12, 45, 7, 23, 89, 54, 3, 18, 90, 31])

print(f"Is array in compatible shapes? {nums1.shape == nums2.shape}")
```

### adding row cols

```python
print(arr)
new_row = np.array([7,8,9])

with_new_row = np.vstack((arr, new_row))

print(with_new_row)
```

```python
original = np.array([[1,2], [3,4]])

new_col = np.array([[56],[89]])

with_new_col = np.hstack((original, new_col))
print(with_new_col)
```

```python
bii = np.arange(6)

delete_elements = np.array([2,4,5])

delete_bii = np.delete(bii, delete_elements)

print(delete_bii)
```

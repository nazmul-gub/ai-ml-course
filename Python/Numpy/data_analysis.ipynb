### Data analysis

```python
import numpy as np

sales_data = np.array([
    [1, 150000, 180000, 220000, 100000], # pizzaburg
    [2, 120000, 140000, 160000, 190000], # pizzahut
    [3, 200000, 230000, 260000, 300000], # dominos
    [4, 180000, 210000, 240000, 270000], # chillox
    [5, 160000, 180000, 200000, 230000] # cha and chill
])

print("Dataset shape:", sales_data.shape)
print("Data type:", sales_data.dtype)
print("First 5 row ", sales_data[:5])
```

```python
# total sales per year
yearly_total = np.sum(sales_data[:, 1:], axis=0)
print(yearly_total)

#minimum sales per retaurent

min_sales = np.min(sales_data[:, 1:], axis=1)
print(min_sales)

#maximum sales per year

max_sales = np.max(sales_data[:, 1:], axis=0)
print(max_sales)

#average sales per year
avg_sales = np.average(sales_data[:, 1:], axis=1)
print(avg_sales)
```

```python
total_sales_per_restaurent = np.cumulative_sum(sales_data[:, 1:], axis=1)
print(total_sales_per_restaurent)
```

```python
vector1 = np.array([1,2,3,4,5])
vector2 = np.array([6,7,8,9,0])

print(vector1+vector2)
print(vector1 * vector2)
print(np.dot(vector1, vector2))

angle = np.arccos(np.dot(vector1, vector2) / (np.linalg.norm(vector1) * (np.linalg.norm(vector2))))
print(angle)
```

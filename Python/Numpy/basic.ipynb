vector, matrix & tensor

```python
vector = np.array([1,2,3])
print(f"vector: {vector}")

matrix = np.array([[1,2,3], [4,5,6]])
print(f"matrix: {matrix}")

tensor = np.array([[[1,2,3], [4,5,6]
                    ,[1,2,3], [4,5,6]]])

print(f"tensor: {tensor}")
```

### shape, dimension, size, dtype

```python
arr = np.array([[1,2,3],
               [4,5,6]])

print(arr)
print(f"Shape: {arr.shape}")
print(f"Dimension: {arr.ndim}")
print(f"Size: {arr.size}")
print(f"Dtype: {arr.dtype}")
print(f"Transpose: {arr.T}")
print(f"Transpose shape: {arr.T.shape}")
```

```python
print(f"reshape array: {arr.reshape(3, 2)}")
print(f"flatten array: {arr.flatten()}")
print(f"ravel array view: {arr.ravel()}")
print(f"floating array: {arr.astype(float)}")
print(f"copy array: {arr.copy()}")
print(f"transpose array: {arr.transpose()}")
print(f"trace: {arr.trace()}")
```

```python
class Phone:
    def __init__(self, model, battery = 100):
        self.model = model
        self.battery = battery

iphone = Phone("17 pro")

# print(iphone.battery)
# print(iphone.model)

class SmartPhone(Phone):
    def __init__(self, model, os, battery=100):
        super().__init__(model, battery)
        self.os = os

    def RunOs(self):
        print(f"{self.os} is running...")

redmi = SmartPhone("Note 14 pro plus", "Oxygen OS")

redmi.RunOs()
print(redmi.model)

class GamingPhone(SmartPhone):
    def __init__(self, model, os, cooling_system, battery=100):
        super().__init__(model, os, battery)
        self.cooling_system = cooling_system

    def start_game(self, name):
        print(f"Playing {name} on {self.model}")

nothing = GamingPhone("Nothing 4 pro", "NOS", "Nytrogen Cooling")
print(nothing.cooling_system)
nothing.start_game("PUBG")


```

```python
class MobilePhone:
    def __init__(self, model, __battery = 100):
        self.model = model
        self.__battery = __battery

    def use(self):
        self.__battery -= 10
        return

    def charge(self):
        if self.__battery >= 100:
            print("Batter is full!")
            return

        self.__battery += 10
        return

    def get_battery(self):
        return self.__battery

nazmul = MobilePhone("Naz 2 pro")

nazmul.use()
nazmul.charge()
nazmul.charge()
print(nazmul.get_battery())
```

Banking Management System 💳

```python
class BankAccount:
    def __init__(self, name, account_number, __balance):
        self.name = name
        self.account_number = account_number
        self.__balance = __balance

    def DisplayInfo(self, account_number):
        print(f"Account Owner: {self.name}\nAccount Number: {self.account_number}\nCurrent Balance: {self.__balance}")
        return

    def Deposit(self, amount):
        self.amount = amount

        min_amount = 500
        max_amount = 500000
        if amount >= min_amount and amount <= max_amount:
            self.__balance += self.amount
        elif amount >= max_amount:
            print("Deposit amount is too much...")
        else:
            print("Deposit amount is too less...")
        return

    def Withdraw(self, amount):
        self.amount = amount
        if amount <= 500000:
            self.__balance -= self.amount
        
        return "Withdraw amount is too much!"

    def GetBalance(self):
        return self.__balance
    

    
```

```python
jahan = BankAccount("Jahan", "245871245jn", 80000)

# jahan.DisplayInfo("245871245jn")
jahan.Deposit(100)

print(jahan.GetBalance())
```

```python
class SavingsAccount(BankAccount):
    def __init__(self, name, account_number, __balance, interest_rate):
        super().__init__(name, account_number, __balance)
        self.interest_rate = interest_rate

    def CalculateInterestRate(self):
        self.interest = (self.__balance * self.interest_rate) / 365

        return self.interest

    def GetTotalBalance(self):
        self.i = self.CalculateInterestRate()
        ans = self.__balance += self.i
        return ans
    
```

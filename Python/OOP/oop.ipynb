## Class and Object

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner            # attribute
        self.balance = balance

    def deposit(self, amount): # method
        self.balance += amount

alice = BankAccount("Alice", 500)
bob = BankAccount("Bob")
alice.deposit(100)
print(alice.balance, bob.balance)
```

### Encaptulation

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.__balance = balance      # private

    @property
    def balance(self):                # read-only view
        return self.__balance

    def withdraw(self, amount):
        if amount <= 0:
            print("Amount must be positive")
        if amount > self.__balance:
            print("Insufficient funds")
        self.__balance -= amount
        print(f"Ok, {amount} is withdrawed")

acct = BankAccount("Alice", 500)
acct.withdraw(100)
print(acct.balance)
# acct.balance = 9999  # can't set the balance, just read the balance
```

### enheritance

```python
class Account:
    def __init__(self, owner, balance = 0):
        self.owner = owner
        self._balance = balance

    def deposit(self, amount):
        self._balance += amount

    @property
    def get_balance(self):
        return self._balance

class SavingsAccount(Account):
    def __init__(self, owner, _balance=0, rate = 0.04):
        super().__init__(owner, _balance)
        self.rate = rate

    def month_end(self):
        self._balance += self._balance * self.rate / 12

class CurrentAccount(Account):
    def __init__(self, owner, _balance=0, overdraft = 500):
        super().__init__(owner, _balance)
        self.overdraft = overdraft

    def month_end(self):
        self._balance -= 5 

jahan = SavingsAccount("Jahan", 500)
jahan.deposit(1000)
jahan.month_end()
print(jahan.get_balance)
```

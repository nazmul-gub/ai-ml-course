### Checking or validating strings

```python
print("123".isdecimal())
print("123".isdigit())
print("123".isnumeric())

print("²".isdecimal())
print("²".isdigit())
print("²".isnumeric())

print("½".isdecimal())
print("½".isdigit())
print("½".isnumeric())
```

### Formatting strings

```python
name = "Nazmul"
score = 95

print("Student: {}, Score: {}".format(name, score))

print(f"Student: {name}, Score: {score}")
```

```python
print("AI".center(8, "-"))  # ---AI---
print("42".zfill(5))        # 00042
print("42".rjust(5, "0"))    # 00042
```

A nice detail about zfill() is that it handles signs:

```python
print("-42".zfill(5))  # -0042
```

### Advanced transformation methods

```python
table = str.maketrans({
    "@": "a",
    "0": "o"
})

text = "N@zmul 0"

print(text.translate(table))
```

```python
table = str.maketrans("", "", "!?")

print("Hello?! Python!".translate(table))
```

```python
text = "Hello"

encoded = text.encode("utf-8")

print(encoded)
print(type(encoded))
```

```python
decoded = encoded.decode("utf-8")

print(decoded)
```

```python
text = "Name\tScore"

print(text.expandtabs(8))
```

```python
# Imagine you have text reviews from an e-commerce dataset.

reviews = [
    "  THIS PRODUCT IS GREAT!!!  ",
    "This product is great",
    "  Great   product   "
]

cleaned_reviews = []

for review in reviews:
    text = review.casefold().strip()
    text = text.replace("!", "")
    text = " ".join(text.split())

    cleaned_reviews.append(text)

print(cleaned_reviews)
```

## set

```python
s = {1, 23, 4, 5}

# type(s)

if 2 in s:
    print("Present")
else:
    print("Not present")
```

```python
s.add(100)
s
```

## mathematical set methods

```python
a = {1, 2, 3, 4}
b = {1, 2}
c = {5, 6, 7}

print(a.union(b))
print(a.intersection(b))
print(a.isdisjoint(c))

print(b.issubset(a))
```

### dict comprehension

```python
square = {x: x**2 for x in range(10)}
square
```

```python
co_ordinates = [(10, 10.5) , (20, 20.5), (30, 30.5)]
locations = ["Patuakhali", "Dhaka", "Lokkhipur"]

exact_locations = {co_or : loc for co_or,loc in zip(co_ordinates, locations)}
exact_locations
```

```python
def adder(a, b):
    """
    added two integer numbers
    a -> int
    b -> int
    """

    sum = a+b
    return sum
```

```python
adder(10, 20)
```

### args

```python
def adders(*args):
    for i in args:
        print(i)

adders(10,20, 30, 40)
```

```python
def add(**kwargs): 
    for key, value in kwargs.items():
        print(f"{key} : {value}")

add(a= 10, b = 20)
```

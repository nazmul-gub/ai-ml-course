### Data inspecting and loading

```python
num_cols = ['units_sold', 'marketing_spend', 'unit_price', 'discount_pct']

for col in num_cols: 
    sns.boxplot(x=df[col])
    plt.title(col + " boxplot")
    plt.show()
```

```python
num_cols = ['units_sold', 'marketing_spend', 'unit_price', 'discount_pct']

for col in num_cols: 
    sns.histplot(x=df[col], kde=True)
    plt.title(col + " boxplot")
    plt.show() 
```

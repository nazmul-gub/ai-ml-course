```python
(df['species'].value_counts() / len(df)) * 100
```

```python
sns.countplot(data=df, x='species')
plt.show()
```

```python
species_value = df['species'].value_counts()

plt.pie(species_value, labels=species_value.index, autopct="%1.1f")
plt.show()
```

```python
(df['island'].value_counts() / len(df)) * 100
```

```python
island_values = df['island'].value_counts()

plt.pie(island_values, labels=island_values.index, autopct="%1.1f")
plt.show()
```

```python
sns.countplot(data=df, x='island')
plt.show()
```

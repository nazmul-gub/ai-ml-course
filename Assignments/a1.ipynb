```python
numbers = list(map(int, input("Enter the integers:").split()))

#4 6 8 12 16 20 24 28
ans = []

for number in numbers:
    if(number % 4 == 0 and number % 8 != 0):
        ans.append(number)

print(ans)
```

```python
def compute_ratio(x, y):
    try:
        return (x + y) / (x * y)
    except:
        return "Error: Division not possible"

x, y = map(int, input().split())
ans = compute_ratio(x, y)
print(ans)
```

Write a program that converts a sentence into a list of words but only includes words that start with a vowel (a, e, i, o, u), ignoring case.


```python
sentence = input().split()

# An apple a day keeps the doctor away

ans = []

for word in sentence:
    if word[0].lower() == "a" or word[0].lower() == "e" or word[0].lower() == "i" or word[0].lower() == "o" or word[0].lower() == "u":
        ans.append(word)

print(ans)
```

### palindrome checker

```python
word = input().strip().lower()

# Radar

reverse_word = word[::-1]

if(reverse_word == word):
    print("Palindrome")
else:
    print("Not Palindrome")
```

Write a Python program that reads a text file and finds the longest word.


```python
with open("data.txt", "r") as file:
    words = file.read().strip().split()
    ans = max(words, key=len)
    print(ans)
```

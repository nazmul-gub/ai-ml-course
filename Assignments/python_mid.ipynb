```python
import numpy as np
import pandas as pd

np.random.seed(24)

num_employees = 50  # you can increase this if needed

# Employee data
emp_ids = np.arange(101, 101 + num_employees)
emp_ages = np.random.randint(20, 55, num_employees)
emp_income = np.random.randint(35000, 95000, num_employees)

# Fixed set of roles (repetition allowed)
roles_list = [
    "HR Manager",
    "Software Developer",
    "Accountant",
    "Marketing Executive",
    "Sales Executive",
    "Backend Developer",
    "HR Executive",
    "Financial Analyst",
    "Digital Marketer",
    "Operations Manager"
]

emp_roles = np.random.choice(roles_list, num_employees)

# Create DataFrame
data = pd.DataFrame({
    "emp_id": emp_ids,
    "emp_age": emp_ages,
    "emp_salary": emp_income,
    "role": emp_roles
})

# Save dataset
data.to_csv("staff_data.csv", index=False)
```

Using .loc:
Increase salary by 12% for "Software Developer" and "Backend Developer"
Increase salary by 5% for all other roles
Then:
Print the dataframe that includes role , previous_salary , new_salary , increase_of_salary


```python
df['new_salary'] = df.loc[df['role'].str.contains("Developer")].iloc[:,2:3] * 1.12
df
```

```python
df['new_salary'] = df.loc[~df['role'].str.contains("Developer")].iloc[:,2:3] * 1.05
```

```python
mask = df['role'].isin(['Software Developer', 'Backend Developer'])

df['previous_salary'] = df['emp_salary']

df.loc[mask, 'new_salary'] = df.loc[mask, 'emp_salary'] * 1.12
df.loc[~mask, 'new_salary'] = df.loc[~mask, 'emp_salary'] * 1.05

df['increase_of_salary'] = df['new_salary'] - df['previous_salary']

df[['role', 'previous_salary', 'new_salary', 'increase_of_salary']]
```
